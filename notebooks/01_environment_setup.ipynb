```python
import os
import random
import numpy as np
import tensorflow as tf

def set_global_seeds(seed=42):
    # 1. Python built-in pseudo-random generator
    random.seed(seed)

    # 2. Python hash seed
    os.environ['PYTHONHASHSEED'] = str(seed)

    # 3. NumPy RNG
    np.random.seed(seed)

    # 4. TensorFlow RNG
    tf.random.set_seed(seed)

    # 5. Enable TensorFlow deterministic ops (prevents cuDNN non-determinism where possible)
    tf.config.experimental.enable_op_determinism()

    print(f"Global seeds set to: {seed}")

set_global_seeds(42)
```
